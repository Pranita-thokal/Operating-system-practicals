```python
import math

process = [2500, 3700, 6100]
page_size = [1024, 2048, 4096]

print("Page Size Analysis")

for size in page_size:
    print("\nPage Size =", size // 1024, "KB")
    total_allocated = 0
    total_waste = 0

    print("Process\tPages\tAllocated\tWaste")

    for p in process:
        pages = math.ceil(p / size)
        allocated = pages * size
        waste = allocated - p

        total_allocated += allocated
        total_waste += waste

        print(p, "\t", pages, "\t", allocated, "\t\t", waste)

    print("Total Allocated =", total_allocated, "bytes")
    print("Total Waste =", total_waste, "bytes")
```
